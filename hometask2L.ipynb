```python
# all valid pairs
all_pairs = []
for a in range(2, 100):
    for b in range(a, 100):
        all_pairs.append((a, b))

# list of pairs and sum, list of pairs manually
prod_to_pairs = {}
sum_to_pairs = {}

for a, b in all_pairs:
    p = a * b
    s = a + b

    if p not in prod_to_pairs:
        prod_to_pairs[p] = []
    prod_to_pairs[p].append((a, b))

    if s not in sum_to_pairs:
        sum_to_pairs[s] = []
    sum_to_pairs[s].append((a, b))

# products with more than one pair
ambiguous_products = []
for p in prod_to_pairs:
    if len(prod_to_pairs[p]) > 1:
        ambiguous_products.append(p)

# find all sums where all pairs lead to ambiguous products
possible_sums = []
for s in sum_to_pairs:
    pairs = sum_to_pairs[s]
    all_ambiguous = True
    for a, b in pairs:
        if (a * b) not in ambiguous_products:
            all_ambiguous = False
            break
    if all_ambiguous:
        possible_sums.append(s)

# get all pairs from these sums
valid_pairs = []
for s in possible_sums:
    valid_pairs += sum_to_pairs[s]

# find products from these valid pairs that point to exactly one pair
product_known = {}
for a, b in valid_pairs:
    p = a * b
    s = a + b

    if s in possible_sums:
        if p not in product_known:
            product_known[p] = []
        product_known[p].append((a, b))

final_product_pairs = []
for p in product_known:
    if len(product_known[p]) == 1:
        final_product_pairs.append(product_known[p][0])

# keep only pairs where the sum is now also unique among remaining ones
final_solution = []
for a, b in final_product_pairs:
    s = a + b
    count = 0
    for x, y in final_product_pairs:
        if x + y == s:
            count += 1
    if count == 1:
        final_solution.append((a, b))
```

```python
# show result
print("The solution is:", final_solution)
```
